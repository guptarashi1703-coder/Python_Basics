```python
def add(*num):
  for i in num:
    print(i)
  return sum(num)
```

```python
def add(*num):
  total=0
  for i in num:
    total+=i
  return total
```

```python
print(add(10,20))
print(add(10,20,30,40,50))
```

**Reshaping Arrays**

```python
import numpy as np
data=np.arange(1,13)
print("Original data shape",data.shape)
a=data.reshape(3,4)
print("3x4 \n",a)
b=data.reshape(4,3)
print("4x3 \n",b)
c=data.reshape(2,-1)
print("2xinfer \n",c)
```

flatten() and ravel()

```python
data=np.array([[1,2,3],[4,5,6]])
print(data.ndim)
flat1=data.flatten()
print("flat1 shape",flat1.ndim)
flat2=data.ravel()
print("flat2 shape",flat2.ndim)
```

```python
data=np.array([[[1,2,3],[4,5,6]]])
print(data.ndim)
flat1=data.flatten()
print("flat1 shape",flat1.ndim)
flat2=data.ravel()
print("flat2 shape",flat2.ndim)
```
