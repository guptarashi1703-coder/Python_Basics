```python
#Q7.3
def print_abs(val):
  if val<0:
    print(-val)
  else:
    print(val)
x=print_abs(-2.7)
print(x)
#The function prints the absolute value (2.7) but does not return anything.In Python, a function without a return statement returns None by default.
```

```python
#Q7.4
def inc_val(val):
   val = val + 1
x = 7
inc_val(x)
print(x)
#Reassigning val inside the function does not change x because val is just a local copy of the argument, and integers are immutable, so modifying val doesn’t affect the original variable x.
```

**Topic 8: Scope — Local vs Global Variables**

```python
#Q8.1
def my_abs(val):
    if val < 0:
        return 0 - val
    return val
print(val)
#This code will raise a NameError because val is defined only as a local parameter inside the function my_abs.
```

```python
#Q8.2
counter=0
def show_counter():
    print(counter)
show_counter()
```

```python
#Q8.3
counter=0
def increment_counter():
    counter=counter+1
    print(counter)
increment_counter()
#Python assumes that counter inside the function is a local variable because you assign to it. Since no local counter exists yet, trying to read it first (counter + 1) causes an UnboundLocalError.
```

```python
#Q8.4
counter=0
def increment_counter():
    global counter
    counter=counter+1
    print(counter)
increment_counter()
increment_counter()
```
